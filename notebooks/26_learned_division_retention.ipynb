```python
from pathlib import Path
import json,warnings
import plotly.graph_objects as go
from IPython.display import Image,display
OUT=Path('/home/sagemaker-user/biohub-cell-tracking-during-development/outputs/competitive_gap_closure/division_retention_cached')
r=json.loads((OUT/'result.json').read_text())
c=json.loads((OUT/'cohort_frozen_before_scoring.json').read_text())
assert r['status']=='complete' and r['network_requests']==0 and r['new_fits']==0

def show(fig,height=570):
    fig.update_layout(width=1040,height=height,margin=dict(l=115,r=35,t=90,b=100))
    with warnings.catch_warnings():
        warnings.filterwarnings('ignore',category=DeprecationWarning)
        png=fig.to_image(format='png',width=1040,height=height,scale=1)
    display(Image(data=png))
print('Decision:',r['decision'])
print('Scored movies:',', '.join(r['measured_movies']))
print('All-eight metric complete:',r['full_eight_video_metric_complete'])
print('No network requests, model fits, or submissions.')
```

```python
a=r['proposal_audit']
fig=go.Figure()
fig.add_bar(name='Exactly two scored daughters',y=[v['stem'] for v in a],x=[v['raw_two_child_parents'] for v in a],orientation='h')
fig.add_bar(name='Both satisfy fixed probability gate',y=[v['stem'] for v in a],x=[v['two_child_above_probability_gate'] for v in a],orientation='h')
fig.update_layout(title='Necessary proposal support in all eight raw prediction graphs',xaxis_title='Parent nodes',yaxis_title='Movie',barmode='group')
show(fig,650)
```

```python
fig=go.Figure()
for arm,label in [('frozen','Frozen reference'),('retained_divisions','Unchanged retention candidate')]:
    s=r['summaries'][arm]
    values=[s['adj_edge_jaccard'],s['division_jaccard'],s['score']]
    fig.add_bar(name=label,x=['Adjusted edge Jaccard','Division Jaccard','Combined local score'],y=values,text=[f'{v:.6f}' if v is not None else 'Not defined' for v in values],textposition='outside')
fig.update_layout(title=f"Measured local comparison — {len(r['measured_movies'])} movie(s), not a leaderboard result",yaxis_title='Pinned organizer-formula metric',barmode='group')
show(fig,550)
```
