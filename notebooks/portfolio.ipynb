## A verified benchmark. An evidence-led research process.

I developed an AWS-based research workflow for reconstructing cell trajectories and
lineages from three-dimensional time-lapse microscopy. The project combines public
reference benchmarking with original representation engineering, controlled graph
experiments, exact error attribution, and tested cloud execution.

**Verified Kaggle public score: 0.947.** This is an officially scored reproduction of
the public Harmonic Fusion inference reference, not a claim that I invented its
pretrained models or surpassed the competition leader.

This portfolio is a **presentation of results and engineering judgment**. Research
source, executable notebooks, exact feature recipes, fitted models, configurations,
and operational workflows are not included in this current presentation snapshot.

## Engineering contribution

**Representation engineering:** implemented and evaluated original spatial,
temporal, intensity, morphology, and candidate-context representations. A controlled
association comparison evaluated 557 numeric features, separating the public-score
features from the original feature contribution.

**Experimental design:** tested alternatives against frozen predictions, used
cross-embryo training/evaluation for learned corrections, and separated official
leaderboard evidence from local exploratory diagnostics.

**Error analysis:** reconciled 111 incorrect and 112 missed edges in five development
movies, including competing assignments, missing detections, and missed divisions.
This directed subsequent work toward limitations the current system could actually address.

**ML engineering:** built bounded AWS execution, artifact-integrity checks, finite-value
validation, recovery receipts, reusable caches, and notebook-output checks. The verified
publication checkpoint passed 277 synthetic tests and CI before and after merge.
These tests establish engineering checks, not predictive generalization.
